```python
import pandas as pd
```

```python
#Define the Customers Dataframe

customers_data = {'id'  : [1, 2, 3, 4],
                  'name': ['Joe', 'Henry', 'Sam', 'Max']}
customers_df = pd.DataFrame(customers_data)

customers_df
```

```python
from typing_extensions import OrderedDict
# Define the Orders DataFrame
orders_data = {'id'        : [1, 2],
               'customerId': [3, 1]}
orders_df = pd.DataFrame(orders_data)

orders_df
```

```python
#Perform a left join to find customers without orders
merged_df = pd.merge(customers_df, orders_df, left_on= 'id', right_on = 'customerId', how = 'left')
merged_df
```

```python
# Filter customers who have no matching order
no_orders_df = merged_df[merged_df['customerId'].isna()]
no_orders_df
```

```python
result = no_orders_df[['name']].rename(columns={'name' : 'Customers'})
print(result)
```
