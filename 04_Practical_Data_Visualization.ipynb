```python
import numpy as np
import pandas as pd
from pandas import Series, DataFrame

from pandas.plotting import scatter_matrix
import matplotlib.pyplot as plt
import seaborn as sns
from pylab import rcParams
```

```python
%matplotlib inline
rcParams['figure.figsize'] = 5, 6
sns.set_style('whitegrid')
```

```python
address = '/content/mtcars.csv'
cars = pd.read_csv(address)
cars.head(3)
```

```python
mpg = cars['mpg']
mpg.plot(kind = 'hist')
```

```python
plt.hist(mpg)
plt.show()
```

# Using Seaborn Lib

```python
sns.displot(mpg)
```

**Seeing Scatterplot in Action**

```python
# Without Sns
cars.plot(kind = 'scatter', x = 'hp', y = 'mpg', c = ['darkgray'])
```

```python
sns.regplot(x = 'hp', y = 'mpg', data = cars, scatter = True)
```

```python
sns.pairplot(cars)
```

**Its a too much information, so lets create a subset of cars and lets plot that**

```python
cars_subset = cars[['mpg', 'disp', 'hp', 'wt']]
sns.pairplot(cars_subset)
plt.show
```

```python
cars.boxplot(column='mpg', by= 'am')
cars.boxplot(column='wt', by= 'am')
```

**Lets do it sns**

```python
sns.boxplot(x = 'am', y = 'mpg', data = cars, palette='hls', hue= 'am')
#sns.boxplot(x = 'am', y = 'wt', data = cars)
```
